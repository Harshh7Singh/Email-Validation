```python
# prompt: generate rectangle design

def rectangle(length, width):
  """Generates a rectangle design using '#' characters.

  Args:
    length: The length of the rectangle.
    width: The width of the rectangle.
  """
  for i in range(width):
    if i == 0 or i == width - 1:
      print("#" * length)
    else:
      print("#" + " " * (length - 2) + "#")

# Example usage:
rectangle(10, 5)
```
